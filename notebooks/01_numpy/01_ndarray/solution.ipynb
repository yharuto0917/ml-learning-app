## ★ 問題 1: `reshape` で 1D → 2D

**解答**: `a.reshape(3, 4)` で行 3 × 列 4 の 2 次元配列に変形します。

**ポイント**: `reshape` は **要素数を変えずに** 形状だけ変える操作。元の要素数 12 = 3 × 4 を満たす形ならどんな組み合わせでも指定できます。

```python
a = np.arange(12)
reshaped = a.reshape(3, 4)

print(reshaped)
```

```python
# --- 自己採点 ---
assert isinstance(reshaped, np.ndarray)
assert reshaped.shape == (3, 4)
assert (reshaped == np.arange(12).reshape(3, 4)).all()
print("✓ 正解")
```

## ★ 問題 2: 行列を直接組み立てる

**解答**: ネストしたリストを `np.array` に渡します。外側のリストが行、内側が各行の列要素になります。

**ポイント**: `np.array([[1, 2], [3, 4], [5, 6]])` のように **外側 = 行・内側 = 列** という対応で書くと shape は `(行数, 列数) = (3, 2)` になります。

```python
M = np.array([[1, 2], [3, 4], [5, 6]])

print(M)
print("shape =", M.shape)
```

```python
# --- 自己採点 ---
assert isinstance(M, np.ndarray)
assert M.shape == (3, 2)
assert (M == np.array([[1, 2], [3, 4], [5, 6]])).all()
print("✓ 正解")
```

## ★★ 問題 3: バグ修正 — 線形変換 `y = W x + b`

**解答**: 行列・ベクトル積は `@` 演算子(あるいは `np.dot`)。`*` は要素ごとの積で、形状が違うとブロードキャストの規則に従って計算されてしまいます。

**ポイント**:
- `*`(要素ごとの積): 同じ shape どうし、もしくはブロードキャスト可能な shape で要素別に掛ける
- `@` / `np.dot`(行列積): 行列とベクトル、行列と行列の数学的な積
- 機械学習で出てくる `Wx`、`Wh`、`xW^T` は全部 `@` 案件

```python
W = np.array([[1, 2], [3, 4]])
x = np.array([5, 6])
b = np.array([1, 1])

y = W @ x + b

print("y =", y)
```

```python
# --- 自己採点 ---
assert y.shape == (2,)
assert (y == np.array([18, 40])).all()
print("✓ 正解")
```

## ★★★ 問題 4: 列ごとの中心化

**解答**: `X - X.mean(axis=0)` の 1 行で書けます。

**ポイント**: ブロードキャストの典型例。
- `X.mean(axis=0)` は **各列の平均** で shape は `(列数,)` = `(3,)`
- shape `(4, 3)` の `X` から shape `(3,)` の値を引くと、NumPy は **(3,) を全行に放送して** `(4, 3)` 形に揃えて減算してくれる
- 結果として「各列の平均が 0」になる行列が手に入る

標準化(平均 0・分散 1)に進む前段としてよく出てくる操作です。

```python
X = np.array([
    [10.0, 1.0, 2.0],
    [20.0, 3.0, 6.0],
    [30.0, 5.0, 10.0],
    [40.0, 7.0, 14.0],
])

X_centered = X - X.mean(axis=0)

print(X_centered)
```

```python
# --- 自己採点 ---
assert isinstance(X_centered, np.ndarray)
assert X_centered.shape == (4, 3)
col_means = X_centered.mean(axis=0)
assert np.allclose(col_means, 0.0, atol=1e-9)
print("✓ 正解")
```

## おつかれさまでした

解答での重要ポイントは次の 3 つです。

1. **`reshape` で形状を変える** — 要素数が一致していれば任意の形に変形可能
2. **`*` と `@` を混同しない** — 線形変換は `@`、要素別計算は `*`
3. **ブロードキャスト** — shape が違っても規則を満たせば自動で形を揃えて演算してくれる
