## ★★ 問題 3: バグ修正 — 線形変換 `y = W x + b`

次のコードはサイト本文の線形変換例ですが、**1 か所バグがあります**。期待される結果 `y = [18, 40]` になるように修正してください。

````python
W = np.array([[1, 2], [3, 4]])
x = np.array([5, 6])
b = np.array([1, 1])

y = W * x + b   # ← ここを直す

print("y =", y)
````

<details><summary>💡 ヒント</summary>

`*` は要素ごとの積です。行列とベクトルの **積** を計算するには `@` または `np.dot` を使います。

</details>

````python
# --- 自己採点 ---
assert y.shape == (2,), f"shape は (2,) を期待。実際: {y.shape}"
assert (y == np.array([18, 40])).all(), f"y = [18, 40] を期待。実際: {y}"
print("✓ 正解")
````
