## 1. セットアップ

Colab には NumPy が標準で入っているのでインストールは不要ですが、念のためバージョンを表示します。

```python
import numpy as np

np.random.seed(0)
print(f"NumPy version: {np.__version__}")
```

## 2. ndarray を作ってみる

Python のリストから 1 次元と 2 次元の配列を作り、`shape`(各次元のサイズ)、`dtype`(要素の型)、`ndim`(次元数)を観察します。

```python
v = np.array([1, 2, 3, 4])
M = np.array([[1, 2], [3, 4], [5, 6]])

print("v       =", v)
print("  shape =", v.shape, "  dtype =", v.dtype, "  ndim =", v.ndim)
print("M       =\n", M)
print("  shape =", M.shape, "  dtype =", M.dtype, "  ndim =", M.ndim)
```

## 3. reshape で形を変える

`reshape` は値はそのままで shape を作り変える操作です。要素数が一致していれば任意の形に変えられます。

```python
a = np.arange(12)        # 0..11 の 1 次元
b = a.reshape(3, 4)      # 3x4 の 2 次元
c = a.reshape(2, 2, 3)   # 2x2x3 の 3 次元

print("a.shape =", a.shape)
print("b =\n", b)
print("b.shape =", b.shape)
print("c.shape =", c.shape)
```

## 4. スカラー演算と配列演算

NumPy では `for` ループを書かずに、配列に対して直接演算できます。これが本書全体で活用される **ベクトル化** の入り口です。

```python
x = np.array([1.0, 2.0, 3.0])

print("x * 2     =", x * 2)
print("x + 10    =", x + 10)
print("x ** 2    =", x ** 2)
print("x + x     =", x + x)
print("x @ x     =", x @ x, "   # 内積")
```

## 5. 線形変換 y = Wx + b

サイト本文で出てきた線形変換を実際に計算してみます。

$$ y = W x + b $$

```python
W = np.array([[1, 2], [3, 4]])
x = np.array([5, 6])
b = np.array([1, 1])

y = W @ x + b
print("y =", y)
```

## まとめ

- `np.array` でリストから配列を作り、`shape` / `dtype` / `ndim` を確認できる
- `reshape` で要素数を保ったまま形を変えられる
- スカラー演算・配列同士の演算が **for ループ無し** で書ける
- 線形変換 `y = W @ x + b` は NumPy 1 行で書ける
